```python
#문자열 출력 방법
print("hello")
print('hello', 'everyone')
print('hello'+'everyone') # + 를 쓰면 붙여서 나오게 된다.
print('hello', end= '!!!') #end를 쓰면 줄바꿈이 안되고 다음 출력 값이 붙여서 나온다
print('test')
```

```python
#변수를 이용한 출력법
a = 123  #숫자
b ='hello' #문자열
print(a, b)
print(a, 456, b, 'world') #456은 숫자열이고 "콤마"는 문자열이다.
```

```python
#ptint 함수 예제-1
print('Hello Python!')
print("Nicr to meet you")
print('Hello "Python"') #쌍따음표가 출력됨
print('Hello' 'Python!')
print('Hello', 'Python!') #콤마는 띄어 쓰기가 되어 출력 됨
print('Hello' + 'Python!') # 공백이 없음
```

```python
#print 함수 예제 -2
print("화면에 직접 출력")
print('ab\'c') #역슬레시를 이용하면 역슬레시 다음에오는 특수문자를 출력가능
print("doesn't")
print('does')
print('doesn\'t')
print("'string'")
print("\"string\"")
print('"string"')
```

```python
#print 함수 예제 -3
s1 = '화면에 직접 출력'
s2 = 'ab\'c'
s3 = 'does'
print(s1)
print(s2)
print(s3)
print(s1[0]) # 첫 글자
print(s2[1]) # 두 번째 글자
print(s3[-1]) # 마지막 글자
#컴퓨터는 0부터 숫자를 세기때문에 1이 두번째 이고 0이 첫번째 이다.
```

```python
#print함수 - 숫자
a = 2  #정수
b = 3.14  #실수
c = a+b
print(a)
print(b)
print(c)
print(round(c, 2)) #소수점 둘째 자리까지 출력  round함수는 반올림 함수 이다.
print("%.2f" % (c)) #소수점 둘째 자리까지 출력
print(a+b)
print(a,b,a+b,c)
d = 1e2  #지수형태의 표현
e = 1e-2
print(d)
print(e)
```

```python
#print함수 혼합
item1 = '사과'
price1 = 1000
item2 = '바나나'
price2 = 500
print(item1, price1)
print(item2, price2)
```

```python
#print함수 혼합
item1 = '사과'
price1 = 1000
item2 = '바나나'
str1 = '{0}는 {1}원입니다.'
print(str1.format(item1, price1))
print(str1.format(item2, price2))
```

```python
item1 = '사과'
price1 = 1000
item2 = '바나나'
price2 = 500
str2 = '%s는 %d원입니다.' # %s인 경우에는 str(문자열)  %d인 경우에는 int 또는 float(정수 또는 실수) 가 올수 있다.
print(str2%(item1, price1))
print(str2%(item2, price2))
```
