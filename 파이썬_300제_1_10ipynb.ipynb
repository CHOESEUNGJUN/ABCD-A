```python
#프린트 줄바꿈
print("first"); print("second")  #프린트문과 프린트문 사이에 세미콜론을 쓰면 프린트를 같은 줄에 써도 줄바꿈이 된다.
```

```python
#프린트 줄바꿈
print("first", end=""); print("second")  # 여기에 end를 붙이면 줄바꿈이 없이 붙여서 출력 된다
```
