```python
!pip install ultralytics --upgrade -q
from ultralytics import YOLO
import cv2
from matplotlib import pyplot as plt
```

```python
# 사전 훈련된 YOLOv8n 모델 (n은 가장 가벼운 버전, 필요시 's', 'm', 'l'로 변경 가능)
model = YOLO('yolov8l.pt')
```

```python
# 이미지 경로 설정
img_path = '/content/1.jpg'  # 자신의 이미지 경로

# 객체 감지
results = model(img_path)

# 시각화 (YOLO 결과는 자동으로 박스를 그린 이미지도 포함함)
results[0].show()  # 박스가 쳐진 이미지를 바로 출력

# 저장된 결과 이미지 보기 (원하면 저장도 가능)
import IPython.display as display
import PIL.Image

result_img = results[0].plot()  # 결과를 NumPy 이미지로
display.display(PIL.Image.fromarray(result_img))
```
