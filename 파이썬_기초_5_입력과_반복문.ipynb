#입력 문법 사용 예제

````python
i = 0
result = 0
while i < 5:
  print (f'{i} 번째', end = " ")
  a = input("성적 입력: ")
  result += int(a)
  i += 1
print(f"평균:  {result/5}")
````

#반복문 - for loop문

```
for 변수 in 리스트 (또는 튜플, 문자열):
  수행할 문장1
  수행할 문장2...
```

````python
test_list = ['one', 'two', 'three']
for i in test_list:
  print(i)
````

```
for 카운터 변수 in
```
(반복 횟수): 반복해서 실행할 명령

````python
for i in range(10):
  print(i,end = " ")
````

````python
result =0
for i in range(1,101):
  result = result + i
  print(f'{i} {result}')
print(result)
````

#for loop문 break

````python
index = 0
s = "BlockDMask"
for a in s:
	print(a, end=' ')
	if a == 'k': #k가 있는 첫번째 위치를 찾아라
		break

	index = index +1 #인텍스의 번호 증가

print(index)
````

#for loop문 contion

````python
student = [180, 170, 164, 199, 182, 172, 177]
for a in student:
  if a > 170:

    continue

  print(a)
````

````python
result = 0
for a in range(1, 101):
	if a % 2 == 0:

		continue
	result = result + a

print(result)
````

#for loop 문 else

````python
l = ['m5' , '911' , 'Rs6']

for name in l:
    if name == '911':
        print('!!GT3 RS!!')
        break
    print(name)
else:
    print('!!FINISH!!')
````

````python
sr = ['father', 'mother', 'brother']
cnt = 0
for s in sr:
        for c in s:
            if c == 'r':
                cnt += 1
print(cnt)
````

````python
a = []

for i in range(3):
    line = []
    for j in range(2):
        line.append(0)
    a.append(line)

print(a)
````

#enumerate() 함수

````python
l = ['Alice', 'Bob', 'Charlie']

for name in l:
    print(name)

for i, name in enumerate(l):
    print(i, name)
````
