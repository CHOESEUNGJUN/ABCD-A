```python
#관계및 논리 연산자
a = 10
b = 3
print("10 == 3 =>", a==b) #크거나 같다
print("10 != 3 =>", a!=b) # 같지 않다.
print("10 > 3 =>", a>b) # 크다
print("10 > =3 =>", a>=b) # 작다
print("10 < 3 =>", a, a <=b) # 작거나 같다
print()
```

```python
#비트연산자 ban(정수)
a = 13
b = 9
print(bin(a)) # 10진수 a를 2진수로 변환
print(bin(b))# 10진수 b를 2진수로 변환
print(bin(a&b)) #비트 and
print(a&b)
```

```python
#비트연산자 or
a = 13
b = 9
print(bin(a))
print(bin(b))
print(bin(a|b)) # 비트 or
print(a|b)
```

```python
#비트연산자 xor
a = 12
b = 8
print(bin(a))
print(bin(b))
print(bin(a^b)) #비트 xor
print('')
print(a^b)
```

```python
#비트연산자
a = 12
b = 8
print(bin(a))
print(bin(~a)) #비트 not
```
